```python
#1.	Load the student marks and attendance data from the provided Excel files into two separate Pandas DataFrames.

attendance=pd.read_excel("Student Performance Record.xlsx",sheet_name="Attendance")
marks=pd.read_excel("Student Performance Record.xlsx",sheet_name="Marks")
print(marks)
print(attendance)

```

```python
#2.	Merge the two DataFrames on the Name column to create a single DataFrame.

Student_record=pd.merge(marks,attendance,on="Name",how="outer")
print(Student_record.head())
```

```python
#3.	Replace attendance values ('Y' or 'N') with numeric values (1 for 'Y' and 0 for 'N').
Student_record[["Attendance Day 1","Attendance Day 2","Attendance Day 3","Attendance Day 4","Attendance Day 5"]]=\
Student_record[["Attendance Day 1","Attendance Day 2","Attendance Day 3","Attendance Day 4","Attendance Day 5"]].replace(to_replace=["Y","N"],value=["1","0"])
print(Student_record.head())

```

```python
from pickle import TRUE
#4.	Deal with the missing values.

'''Student_record.info()
Student_record.isna().sum()
print((Student_record == " ").sum()'''


Student_record.replace(to_replace=[" "],value=0,inplace=True)
Student_record.fillna("0",inplace=True)
print(Student_record.head(30))

#check if all the null and empty spaces are taken care of
print(Student_record.info())
print(Student_record.isna().sum())
print((Student_record == " ").sum())
```

```python
#5.	Do the column cleaning like name column is the combination of upper and lower case.


Student_record["Name"]=Student_record["Name"].str.capitalize()
Student_record["Name"]=Student_record["Name"].str.strip()
Student_record.columns.str.strip()
Student_record.head(10)
```

## Data Transformation

```python
print(Student_record.loc[:,["Mini Test 1","Mini Test 2",'Live Test',"Assignment"]].astype(int).max())
print(Student_record.loc[:,["Mini Test 1","Mini Test 2",'Live Test',"Assignment"]].astype(int).min())
```

```python
Student_record[Student_record["Live Test"].astype(int) >15]
```

```python
Student_record.loc[95,"Live Test"]=13 #considering 113 as type and replacing it with 13
```

```python
#o	Total marks for each student.
Student_record["Marks_obtained"]=Student_record.loc[:,["Mini Test 1","Mini Test 2",'Live Test',"Assignment"]].sum(axis=1)
```

```python
#Percentage marks for each student
Total_marks=55
Student_record["Percentage"]=((Student_record.loc[:,"Marks_obtained"].astype(int)/Total_marks)*100).round(1)

Student_record.head()
```

```python
#o	Attendance percentage for each student based on attendance columns
Student_record["Attendance"]=((Student_record.loc[:,"Attendance Day 1":"Attendance Day 5"].astype(int).sum(axis=1)/5)*100).round()
```

```python
Student_record.head()
```

```python
#o	Calculate the weighted percentage – Attendance(40%), Mini test01(10%), Mini test02(10%), Live test(20%), Assignment (20%).
Student_record["Weighted_Percentage"]=(
    Student_record["Attendance"].astype(int) *0.4 +
    ((Student_record["Mini Test 1"].astype(int)/10)*100) * 0.1 +
    ((Student_record["Mini Test 2"].astype(int)/10)*100) * 0.1 +
    ((Student_record["Live Test"].astype(int)/15)*100) * 0.2 +
    ((Student_record["Assignment"].astype(int)/20)*100) * 0.2
).round()
```

```python
Student_record.head(20)
```

# 2.	Classify each student's performance into categories:

```python
def categorize_percentage(percentage):
    if percentage >= 85:
        return "Excellent"
    elif 71 <= percentage < 85:
        return "Good"
    elif 50 <= percentage < 71:
        return "Average"
    else:
        return "Needs Improvement"
```

```python
Student_record["Performance Category"] = Student_record["Weighted_Percentage"].apply(categorize_percentage)
```

# Visualization

```python
#1.	Create a bar chart displaying weighted percentages for top 5 students.
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.pyplot as plt


top_students = Student_record.sort_values(by="Weighted_Percentage", ascending=False).head(5)


plt.figure(figsize=(10,6))
plt.bar(top_students["Name"], top_students["Weighted_Percentage"], color='skyblue')
plt.title('Top 5 Students by Weighted Percentage')
plt.xlabel('Students')
plt.ylabel('Weighted_Percentage')
plt.xticks(rotation=45)
plt.show()
```

```python
#2.	Create a pie chart showing the distribution of students across the four performance categories.

category_counts = Student_record["Performance Category"].value_counts()


plt.figure(figsize=(7,7))
category_counts.plot.pie(autopct='%1.1f%%', colors=["lightblue", "lightgreen", "orange", "red"], startangle=90)
plt.title('Distribution of Students Across Performance Categories')
plt.ylabel('')
plt.show()
```

```python
#3.	Create box plots for each test (Live Test, Mini Test 1, Mini Test 2, Assignment) to visualize the spread and detect potential outliers in scores.

import plotly.express as px

# Create a box plot using Plotly
fig = px.box(
    Student_record,
    y=["Live Test", "Mini Test 1", "Mini Test 2", "Assignment"],
    title="Box Plot for Test Scores",
    labels={"value": "Scores", "variable": "Test"}
)


fig.show()


```

```python
#4.	Create a chart to show the students where attendance is less than 50%.


low_attendance = Student_record[Student_record["Attendance"] < 50]

plt.figure(figsize=(10, 6))
sns.barplot(
    data=low_attendance,
    x="Name",
    y="Attendance",
    palette="Blues_d"
)
plt.title("Students with Attendance < 50%")
plt.xlabel("Students")
plt.ylabel("Attendance (%)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()

```

```python
#5.	Any other visualization/analysis which you can infer from data for the management.


corr_matrix = Student_record[["Attendance", "Mini Test 1", "Mini Test 2", "Live Test", "Assignment", "Weighted_Percentage"]].corr()

plt.figure(figsize=(10,6))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlation Heatmap of Attendance and Test Scores")
plt.show()
```
