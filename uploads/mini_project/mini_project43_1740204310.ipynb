```python
#Question 1
import numpy as np
array1 = np.array([1, 2, 3, 0])
array2 = np.array([4, 1, 0, 6])

addition = array1 + array2
subtraction = array1 - array2
multiplication = array1 * array2
division = array1 / array2

print("Addition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
print("Division:", division)
```

```python
#Question 2
array1 = np.random.randint(1, 101, size=5)

array2 = np.random.randint(1, 101, size=(4, 5))

addition = array2 + array1  
subtraction = array2 - array1 
print("Array 1 (1-D):", array1)
print("Array 2 (2-D):")
print(array2)
print("Addition (Array 2 + Array 1):")
print(addition)
print("Subtraction (Array 2 - Array 1):")
print(subtraction)
```

```python
#Question 6
#Stacking in NumPy refers to the process of joining multiple arrays along a new axis. It allows you to combine arrays in a way that results in a higher-dimensional array.

import numpy as np

# Create two 1-D arrays
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
stacked = np.stack((a, b))
print("Stacked Array (np.stack):")
print(stacked)
```

```python
#Question 6 Difference between the two
# 1. Stacking: Joins arrays along a new axis, increasing dimensionality.
Concatenation: Joins arrays along an existing axis, keeping the same dimensionality.
# 2. Stacking uses functions like np.stack, np.vstack, np.hstack, and np.dstack.
Concatenation uses the np.concatenate function.
# 3. Stacking can increase the dimensionality of the result. For example, stacking two 1-D arrays results in a 2-D array.
Concatenation keeps the same number of dimensions. For example, concatenating two 1-D arrays will still yield a 1-D array
```
