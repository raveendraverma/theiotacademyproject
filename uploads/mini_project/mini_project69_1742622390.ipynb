```python
Library={"book1":{"author":"Test1","Available":True,"Available":True},"book2":{"author":"Test2","Available":True},"book3":{"author":"Test3","Available":True}}
Library
```

```python
#display function
def display(Library):
  for book_name,book_details in Library.items():
    print("\n",(book_name,Library[book_name]['author'],Library[book_name]['Available']))





```

```python
def add_books():

  while True:
    new_book=input("\n\nenter book name: or 'e' to exit ").lower()
    if new_book=="e":
      break
    if new_book not in Library:
        author_name=input("Enter Author Name: ").lower()
        Library[new_book]={"author":author_name,"Available":True}
        print(f"{new_book} added to the Library")

    else :
       print("Book already exist")



```

```python
#issue records
issue_record={}

def issuing_books():

  while True:
    try:
      Choose_option=int(input("\n\n    Choose option: \n1.Issue Book\n2.Exit "))
      if Choose_option in [1,2]:
        if Choose_option==1:
          issuing_book=input("Enter Book Name: ").lower()
          if (issuing_book in Library) and (Library[issuing_book]['Available']==True):

            issuer_name=input("Enter issuer name : ").lower()
            issue_record[issuing_book]=issuer_name
            Library[issuing_book]['Available']=False
            print(f"{issuing_book} is issued to {issuer_name}\n\n")



          else:
            if (issuing_book in Library):
              if Library[issuing_book]['Available']==False :
                print(f"{issuing_book} is already issued")
            else:
              print(f"{issuing_book} is not in Library")
      else:
        print("\nWrong choice\n")

      if Choose_option==2:
          break

    except ValueError:
      print("Invalid Input")








```

```python
def view_issued_books():
  print("\nBook name   Issued to")

  if len(issue_record)!=0:
    for i,k in issue_record.items():
        print("  ",i,"    ",k)
  else:
    print("No books have been issued")



```

```python
def returning_books():
  return_book=input("Enter Book name to return :")
  while True:
    if return_book in issue_record:
      issue_record.pop(return_book)
      Library[return_book]['Available']=True
      print(f"{return_book} is added to the library\n\n")
      break


    else:
      print("Book is not issued by library")
      break






```
