```python
import re
from robobrowser import RoboBrowser

from bs4 import BeautifulSoup as bs

def getData(from_station, to_station):
    br=RoboBrowser(history=True)
    br.open("http://www.delhimetrorail.com/metro-fares.aspx")
    form=br.get_form()
    form['ctl00$MainContent$ddlFrom'].value=from_station
    form['ctl00$MainContent$ddlTo'].value=to_station
    br.submit_form(form,submit=form['ctl00$MainContent$btnShowFare'])
    src=str(br.parsed())
    return src

if __name__ == '__main__':

    st=getData("HUDA CITY CENTRE","ANAND VIHAR")
    nf= st.index('class="fare_new_nor_right"')
    print("Normal Fare: ",st[nf+27:nf+29])

    cf=st.index('class="fare_new_right_right"')
    print("Concessional Fare: ", st[cf + 29:cf + 31])
    print("cf=",cf)
    tm = st.index('Timing -')
    tm1=st[tm + 9:tm + 13]
    for s in tm1.split():
        if s.isdigit():
            print("Timings: ",s)


    stn = st.index('Stations -')
    print("Stations: ", st[stn + 11:stn + 13])

    inter = st.index('Interchange -')
    print("Interchange: ", st[inter + 14:inter + 15])
```

## Special/Magic Method -> Constructor

```python
class Human:
    
    #class attribute
    species = 'Mammal'
    
    # Special Method -> Constructor
    # This method gets invoked whenever an object is created
    def __init__(self):
        print("Human Created")
        
    # General Method    
    def say_hello(self):
        print("Hello")
        
a = Human()
b = Human()

a.say_hello()
```

```python
class Human:
    
    #class attribute
    species = 'Mammal'
    
    # Special Method -> Constructor
    # This method gets invoked whenever an object is created
    def __init__(self,name,age):
        print("Human Created")
        # Object/Instance Attribute
        self.name = name
        self.age = age
        
    # General Method    
    def say_hello(self):
        print("Hello")
        
a = Human('Rishabh',28)
b = Human('Raj',27)

a.say_hello()
print(a.name,a.age)
```

## Creating a Circle class

```python
class Circle:
    
    # Class Attribute
    pi = 3.14
    
    # Constructor
    def __init__(self,radius=1):
        self.radius = radius
        
    # Methods
    def area(self):
        return Circle.pi * self.radius **2
    
    def circumference(self):
        return 2 * Circle.pi * self.radius
    
c1 = Circle()
print(c1.radius)
print(c1.area())
print(c1.circumference())

c2 = Circle(10)
print(c2.radius)
print(c2.area())
print(c2.circumference())
```

## Class Rectangle

```python
class Rectange:
    
    def __init__(self,length,width):
        self.length = length
        self.width = width
        
    def area(self):
        return self.length * self.width
    
    def perimeter(self):
        return 2*(self.length + self.width)
    
r = Rectange(10.0,20)
print(r.area())
print(r.perimeter())
```
