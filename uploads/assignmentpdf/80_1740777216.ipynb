### Q5 Write a Function to check if the year number is a leap year or not. Print all Leap Years of 21st Century.

```python
Leap_years = []
print("The leap years of 21st Century:")
for i in range(2001, 2100):    
    if (i % 4 == 0 and i % 100 != 0) or (i % 400 == 0):
        Leap_years.append(i)
        print(i)
print("Total number of leap years :", len(Leap_years))
```

```python
def longest_word(x):
    words = x.split()
    Longest_word = ""
    length = 0
    for i in words:
        if (len(i)> length):
            Longest_word = i
            length = len(i)
            continue
        return Longest_word
```

### Q10. Write a Python  program  that takes a list and counts  the frequency of each  element using a dictionary. The keys of the dictionary should be the list elements, and the values should be the counts

```python
input_list = [1,2,2,3,4,4,4,5,5]
output_dict = {}
for i in input_list:
    if i in output_dict:
        output_dict[i] += 1
    else:
        output_dict[i] = 1
print(output_dict)
```
