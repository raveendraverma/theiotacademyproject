```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
file_path = "IPL_Dataset.xlsm"
df = pd.read_excel(file_path, sheet_name="IPL_Dataset")

print("\nQ1. What is the maximum number of matches played by an individual player in a season? Print the player name along with the number of matched played.")
top_player = df[df["Matches"] == df["Matches"].max()][["Player", "Matches"]]

print(top_player)

print("\n Q2. Top 2 players with maximum Average who have scored atleast 2 half centuries ?")
top_2_avg = df[df[50] >= 2].nlargest(2, "Avg")[["Player", "Avg", 50]]
print(top_2_avg)

print("\n Q3. Create 2 new columns based on Player name. First column will have first name and second column will have last name. Eg: for the player Shikhar Dhawan, Shikhar will be the first name and Dhawan will be the last name.")
df[["First Name", "Last Name"]] = df["Player"].str.split(" ", n=1, expand=True)

print(df)


print("\nQ4. Create a new column (Cleaned_Highest_score) based on Highest score variable. Remove the Asterik(*) mark and convert the data type into INT.")
df["Cleaned_Highest_score"] = df["Highest Score"].astype(str).str.replace("*", "").replace("nan", "0").astype(float).astype(int)

print(df)

print("\nQ5. Print the total number of centuries scored in the entire season.")
total_centuries = df[100].sum()
print(total_centuries)

print("\nQ6. Print all the player names whose strike rate is less than the average strike rate of all players in entire season. Print the player name, his strike rate and average strike rate.")
avg_strike_rate = df["Strike rate"].mean()
df["Diff with AVG SR"] = df["Strike rate"].mean()-df["Strike rate"]
low_strike_players = df[df["Strike rate"] < avg_strike_rate][["Player", "Strike rate","Diff with AVG SR"]]
df_sorted = low_strike_players.sort_values(by="Strike rate", ascending=True)

print(df_sorted)

print("\nQ7. Please check the correlation between the features and create a heat map.")
plt.figure(figsize=(10,6))
sns.heatmap(df.select_dtypes(include=['number']).drop(columns=["Jersey No"]).corr(), annot=True, cmap="coolwarm")
plt.title("Feature Correlation Heatmap")
plt.show()

print("\nQ8. Check the list of players who has an average greater than 50 as well strike rate above 120. Print player name, average and strike rate.")
high_perf_players = df[(df["Avg"] > 50) & (df["Strike rate"] > 120)][["Player", "Avg", "Strike rate"]]
print("\nQ8:", high_perf_players)

print("\nQ9. Please check the list of players who has an average greater than 40 and balls faced above 100. Print player name, average and balls faced.")
consistent_players = df[(df["Avg"] > 40) & (df["Balls faced"] > 100)][["Player", "Avg", "Balls faced"]]
print("\nQ9:", consistent_players)

print("\nQ10. Players who scored atleast one century in this season. Create visualization.")
plt.figure(figsize=(8,5))
sns.barplot(x=df[df[100] >= 1]["Player"], y=df[df[100] >= 1][100])
plt.xticks(rotation=90)
plt.title("Players with at least one century")
plt.show()

print("\n Q11. Players who scored atleast 4 half centuries in this season.")
print("\nQ11:",df[df[50] >= 4][["Player", 50]].sort_values(by=50, ascending=False))


print("\n Q12. Check the list of players who hit more than 45 boundaries and more than 10 sixes in this season.")
boundary_six_players = df[(df["4s"] > 45) & (df["6s"] > 10)][["Player", "4s", "6s"]]
print("Q12:", boundary_six_players)

print("\n Q13. Plot a histogram of number of matches played in a season by players.")
plt.hist(df["Matches"], bins=10, edgecolor="black")
plt.title("Histogram of Matches Played")
plt.xlabel("Matches")
plt.ylabel("Count of Players")
plt.show()

print("Q14. Plot the histogram of balls faced by players.")
plt.hist(df["Balls faced"], bins=10, edgecolor="black")
plt.title("Histogram of Balls faced")
plt.xlabel("Balls Faced")
plt.ylabel("Count of Players")
plt.show()

print("Q15. Top 10 players with most runs in a season.")
print("\nQ15:", df.nlargest(10, "Runs")[["Player", "Runs"]])

print("Q16. Print the players who played the match but didn't get the batting.")
non_batting_players = df[df["Inns"] == 0]["Player"]
print("\nQ16:", non_batting_players)

print("Q17. Create a new column to show the percentage of total runs scored in 4s and 6s. Then print the top 5 players with maximum percentage.")
df["Boundary Percentage"] = ((df["4s"] * 4 + df["6s"] * 6) / df["Runs"]) * 100
top_5_boundary_pct = df.nlargest(5, "Boundary Percentage")[["Player", "Boundary Percentage"]]
print("Q17:", top_5_boundary_pct)

print("Q18. Print the players with top 5 Not out percentages (Not Out percentage can be calculated as number of Not outs divided by Innings).")
df["Not Out Percentage"] = df["Not Out"] / df["Inns"]*100
top_5_not_out = df.nlargest(5, "Not Out Percentage")[["Player", "Not Out Percentage"]]
print("Q18:", top_5_not_out)

print("\n Q19. Create visualization of top 10 players with highest number of sixes.")
plt.figure(figsize=(8,5))
sns.barplot(x=df.nlargest(10, "6s")["Player"], y=df.nlargest(10, "6s")["6s"])
plt.xticks(rotation=90)
plt.title("Top 10 Players with Most Sixes")
plt.show()

print("Q20. Scatter plot of runs scored by a player v/s balls faced in a season. Then find the relationship between these 2 variables.")
print('\n')
plt.figure(figsize=(8,5))
sns.scatterplot(x=df["Balls faced"], y=df["Runs"])
plt.xlabel("Balls Faced")
plt.ylabel("Runs Scored")
plt.title("Runs Scored vs Balls Faced")
plt.show()
```
