```python
#g)	Delete all the rows where Director column has missing values.
Netflix_data.replace([" ",""],np.nan,inplace=True)

DroppedNAN=Netflix_data.dropna(axis=0,subset=["director"])
print(DroppedNAN.count(axis=0))
print(Netflix_data.count(axis=0))
```

```python
#h)	Print all the records where country has Germany value (including West Germany). If any other country is there along with Germany, then that row should also come in output.
Germany_Records=Netflix_data[Netflix_data["country"].str.contains("germany|west germany",case=False,na=False)]
print(Germany_Records)
```

```python
#i)	Expand Duration column into 2 separate columns – First column having the numeric value and other having String. Eg: 3 seasons should be split in 2 columns having 3 in 1st column and seasons in 2nd column.
Netflix_data[["Duration_No", "Duration_String"]] = Netflix_data["duration"].apply(lambda x: pd.Series(str(x).split(" ", 1)) if pd.notna(x) else pd.Series([np.nan,np.nan]))
print(Netflix_data.sample())
```

```python
#j)	Split Date added into 3 separate columns having date value in 1st column, month value in 2nd column and year value in 3rd.

Netflix_data["date_added"]=pd.to_datetime(Netflix_data["date_added"],format="mixed")

Netflix_data["day"] = Netflix_data["date_added"].dt.day
Netflix_data["month"] = Netflix_data["date_added"].dt.month_name()
Netflix_data["year"] = Netflix_data["date_added"].dt.year

print(Netflix_data.head(1))

```

```python
#k)	Print the number of TV shows/Movies released in each year
print(Netflix_data.groupby(["year","type"]).size())

```

```python
#l)	Rename the column title with movie_title.
Netflix_data.rename(columns={"title":"movie_title"},inplace=True)
print(Netflix_data.head(5))
```

```python
#m)	Split Listed_in column into 3 different columns with col name (Genre1, Genre2, Genre3). Split the column based on comma.
Netflix_data[["Genre1","Genre2","Genre3"]]=Netflix_data["listed_in"].str.split(",",expand=True)
print(Netflix_data.head())
```

```python
#e)	Create a new Column ‘IQ_level’ which will have 3 values (Intelligent, Mediocre, weak). If student scored than 80 then Tag him as Intelligent, if student scored between 50-80, then Mediocre, else weak.

def scores(score):
  if score > 80:
    return "Intelligent"
  elif score in range(50,81):
    return "Mediocre"
  else :
    return "Weak"

Student_info["IQ_Level"]=Student_info["Mark"].apply(scores)

print(Student_info.sample(5))
```

```python
'''3.	Explain the concept of missing values. How can you identify the missing values in a Pandas DataFrame ?
What are the different ways of treating/Imputing/Deleting the missing values.
Explain with example.'''

'''Ans. Missing values are the values that were not recorded or not entered in the data frame
when the data is imported in pandas it leads to creation of Nan or None values at the missing value loacations,
NaN for intger/float while None is for object type data.

To identify missing values in NaN/None we us df.isna() or df.isnull() function along with .sum() to count the number of
find number of missing values in the data frame.isna() and isnull() function returns True for missing values and False for
for non-missing values.To get the overview of the data we can use df.info(),it returns summary of the data frame along with
on_null value count.

Treating null values:

1.Deleting NaN values:
  a.drop(axis=0)-->drops rows containg missing values
  b.drop(axis=1)-->drops columns containing missing values
  c.drop()-->drops missing values

2.Imputting Values: Imputation refers to filling in missing values with a
  specific value, mean, median, mode, or even predictions from a machine learning model.
  Function: a)fillna() function is used to fill missing value
            instead of constand value aggregation can be done(such as medium,mean etc.)
            b)ffill() and bfill() can be used to replace the missing value with next observation
            or last observation.

Eg.
```

```python
import pandas as pd
import numpy as np

# Example DataFrame
data = {'Name': ['Alice', 'Bob', 'Charlie', np.nan],
        'Age': [25, 30, np.nan, 22],
        'Gender': ['Female', np.nan, 'Male', 'Male']}
df = pd.DataFrame(data)

# Identify missing values
print("Missing values:")
print(df.isna().sum())

# 1. Drop rows with missing values
df_dropped = df.dropna()

# 2. Fill missing values with the mean for 'Age' and mode for 'Gender'
df_filled = df.copy()
df_filled['Age'] = df_filled['Age'].fillna(df_filled['Age'].mean())
df_filled['Gender'] = df_filled['Gender'].fillna(df_filled['Gender'].mode()[0])

# 3. Forward fill for missing values in 'Gender'
df_filled_ffill = df.fillna(method='ffill')


print("\nAfter dropping missing values:")
print(df_dropped)

print("\nAfter filling missing values with mean for 'Age' and mode for 'Gender':")
print(df_filled)

print("\nAfter forward filling missing values:")
print(df_filled_ffill)

```
