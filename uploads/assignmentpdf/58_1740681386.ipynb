3.Explain the concept of missing values. How can you identify the missing values in a Pandas DataFrame ?
What are the different ways of treating/Imputing/Deleting the missing values.
Explain with example.


Missing values refer to the absence of data in a dataset. 
It can arise due to various reasons such as data entry errors, data collection issues, or information is not available int the dataset.

How to dentify the missing values:
Missing values can be identified in a Pandas Dataframe by using functions isna() or isnull().
These methods return a DataFrame of the same shape, where each element is a boolean value indicating if the element is missing (True) or not (False).

How to treating-Missing Values :
There are several ways to handle missing values, depending on the context and the importance of the data:
1. dropna(): This method can be used to drop rows or columns with missing values
2. fillna(): This method can be used to fill missing values with specified values, such as mean, median, or mode.

```python
######Example :
# Sample DataFrame
data = {
    'Employee_ID': [1, 2, 3, 4, 5],
    'Name': ['John', 'Jane', None, 'Mike', 'Anna'],
    'Age': [28, None, 35, 40, None],
    'Department': ['HR', 'Finance', 'IT', None, 'Marketing']
}

df = pd.DataFrame(data)
print("Original DataFrame:\n", df)
# Identifying missing values
missing_values = df.isna()
print("\nMissing Values:\n", missing_values)

# Summary of missing values in each column
missing = df.isna().sum()
print("\nSummary of Missing Values:\n", missing)

# Dropping rows with any missing values
df_dropped_rows = df.dropna()
print("\n Dropping Rows with Missing Values:\n", df_dropped_rows)

# Dropping columns with any missing values
df_dropped_cols = df.dropna(axis=1)
print("\nDropping Columns with Missing Values:\n", df_dropped_cols)

# Filling missing values with mean (for numerical data)
df['Age'].fillna(df['Age'].mean(), inplace=True)
print("\nDataFrame after Filling Missing Age with Mean:\n", df)

# Filling missing values with a specific value
df['Name'].fillna('unknown', inplace=True)
print("\nDataFrame after Filling Missing Names with 'unknown':\n", df)
```
