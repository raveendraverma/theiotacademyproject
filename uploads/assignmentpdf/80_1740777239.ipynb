### Q1.Create the first numpy array of 1-D with values (1,2,3,0).
###  Create a second numpy array of 1-D with values (4,1,0,
###  6). Perform following simple arithmetic operations on
###  these two numpy arrays:
###  • Addition
###  • Subtraction
###  • Multplication
###  • Division

```python
arr1 = np.array([1, 2, 3, 0])
arr2 = np.array([4, 1, 0, 6])
addition = arr1 + arr2
subtraction = arr1 - arr2
multiplication = arr1 * arr2

with np.errstate(divide='ignore', invalid='ignore'):
    division = arr1 / arr2 
    division[np.isnan(division)] = 0 

print("Addition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
print("Division:", division)
```

### Q4. Create a 2-D numpy array of random integers between
###  1 to 100 with shape 5*4. Now perform following
###  operations:
###  • Print only the even numbers from this array
###  • Print only the odd numbers from this array
###  • Print row wise sum of all elements
###  • Print column wise sum of all elements
###  • Convert this 2-D array into 3-D array of shape
###  2*2*5 

```python
import numpy as np

arr_2D = np.random.randint(1, 101, (5, 4))
even = arr_2D[arr_2D % 2 == 0]
odd = arr_2D[arr_2D % 2 != 0]
row_wise_sum = np.sum(arr_2D, axis=1)
column_wise_sum = np.sum(arr_2D, axis=0)
arr_3D = arr_2D.reshape(2, 2, 5)

print("Original 2-D Array:\n", arr_2D)
print("\nEven Numbers:\n", even)
print("\nOdd Numbers:\n", odd)
print("\nRow-wise Sum of Elements:\n", row_wise_sum)
print("\nColumn-wise Sum of Elements:\n", column_wise_sum)
print("\n3-D Array (2x2x5):\n", arr_3D)
```

### Q5. Create a 2-D numpy array of random integers between
###  1 to 100 with shape 5*4. Now perform following
###  operations:
###  • Swap first row with third row
###  • Swap second column with fourth column.
###  • Replace the values less than 50 with zero.
###  • Convert this 2D array into 1-D array

```python
arr_2D = np.random.randint(1, 101, (5, 4))
arr_2D[[0, 2]] = arr_2D[[2, 0]]
arr_2D[:, [1, 3]] = arr_2D[:, [3, 1]]
arr_2D[arr_2D < 50] = 0
arr_1D = arr_2D.flatten()

print("Array after swapping first row with third row:\n", arr_2D)
print("\n2-D Array second column with fourth column:\n", arr_2D)
print("\n2-D Array with Values Less Than 50 Replaced by Zero:\n", arr_2D)
print("\nConverted 1-D Array:\n", arr_1D)
```

### Q6. Explain the concept of stacking in numpy with examples.
###  How is it different from concatenation. Give any 3
###  differences between them.


##### Stacking helps in joining multiple arrays. It joins arrays of same dimension. Stacking is done alone new axis.
##### 3 differences between stacking and concatenation;
##### i. Concatenation joins multiple arrays but along an existing axis.
##### ii.  It does not create a new dimension.
##### iii. Function used is np.concatenate().

```python
# Examples of stacking
# 1. Vertical stacking

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

result = np.vstack([arr1, arr2])
print("Vertical Stacking:\n", result)
```

```python
# 2. Horizontal stacking

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

result = np.hstack([arr1, arr2])
print("Horizontal Stacking:\n", result)
```

```python
# Depth stacking

arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8]])

result = np.dstack([arr1, arr2])
print("Depth Stacking (dstack):\n", result)
```
