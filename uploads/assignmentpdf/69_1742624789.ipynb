```python
'''Q4. Create a 2-D numpy array of random integers between
 1 to 100 with shape 5*4. Now perform following
 opera�ons:
 • Print only the even numbers from this array
 • Print only the odd numbers from this array
 • Print row wise sum of all elements
 • Print column wise sum of all elements
 • Convert this 2-D array into 3-D array of shape
 2*2*5'''

arr=np.random.randint(1,100,(5,4))
print(arr[arr%2==0]) #Print only the even numbers from this array
print(arr[~arr%2==0]) #Print only the odd numbers from this array
print(np.sum(arr,axis=1)) #Print row wise sum of all elements
print(np.sum(arr,axis=0)) #Print colomn wise sum of all elements
Arr_3D=arr.reshape(2,2,5)
print(Arr_3D)


```

```python
'''Q5. Create a 2-D numpy array of random integers between
 1 to 100 with shape 5*4. Now perform following
 opera�ons:
 • Swap first row with third row
 • Swap second column with fourth column.
 • Replace the values less than 50 with zero.
 • Convert this 2D array into 1-D array'''

arr=np.random.randint(1,100,(5,4))
arr[[0,2]]=arr[[2,0]] #Swap first row with third row
arr[:,[1,3]]=arr[:,[3,1]] #Swap second column with fourth column.
arr[arr<50]=0 #Replace the values less than 50 with zero.
np.flatten(arr) #Convert this 2D array into 1-D array


```
