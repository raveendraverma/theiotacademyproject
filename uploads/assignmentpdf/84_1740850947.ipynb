```python
#3.Explain the concept of missing values. How can you identify the missing values in a Pandas DataFrame ?
#What are the different ways of treating/Imputing/Deleting the missing values.
#Explain with example.
```

Missing values in a dataset occur when no data is stored for a particular variable in an observation. These values can arise due to various reasons, such as human error, data corruption, or system limitations.

Identifying Missing Values - 
1. Using isna() or isnull()
2. Using sum() with isna() to count missing values
3. Using info()

Different Ways to Handle Missing Values -
1. Deleting Missing Values
--> Using dropna(): Removes rows or columns containing missing values
    df_cleaned = df.dropna()  # Removes rows with missing values
    print(df_cleaned)
--> Removing Columns with Many Missing Values :
   df_dropped_columns = df.dropna(axis=1)

2. Imputing Missing Values
     a) Filling with a Specific Value (fillna()) :

       df_filled = df.fillna(0)  # Replaces all NaN values with 0
       print(df_filled)
     b) Filling with Mean, Median, or Mode
     c) Using Interpolation : Interpolation estimates missing values based on available data.
     d) Using SimpleImputer from Scikit-Learn (for Machine Learning)

```python
import pandas as pd

data = {
    'Product_ID': [101, 102, 103, 104, 105],
    'Product_Name': ['Laptop', 'Mobile', None, 'Tablet', 'Camera'],
    'Price': [75000, None, 30000, 25000, None],
    'Category': ['Electronics', 'Electronics', 'Electronics', None, 'Electronics']
}

df = pd.DataFrame(data)
print("Original DataFrame:\n", df)

missing_values = df.isna()
print("\nMissing Values:\n", missing_values)

missing = df.isna().sum()
print("\nSummary of Missing Values:\n", missing)

df_dropped_rows = df.dropna()
print("\nDropping Rows with Missing Values:\n", df_dropped_rows)

df_dropped_cols = df.dropna(axis=1)
print("\nDropping Columns with Missing Values:\n", df_dropped_cols)

df['Price'].fillna(df['Price'].mean(), inplace=True)
print("\nFilling Missing Price with Mean:\n", df)

# Filling missing values with a specific value
df['Product_Name'].fillna('Unknown', inplace=True)
print("\nFilling Missing Product Names with 'Unknown':\n", df)
```
