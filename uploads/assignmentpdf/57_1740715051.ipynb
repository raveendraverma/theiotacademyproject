```python
#Already imported pandas for the 1st question 

#Imported the csv files
students_df = pd.read_csv('student.csv')
marks_df = pd.read_csv('marks.csv')

print("\nA: Print the column names to check for any discrepancies")
print("Students DataFrame Columns:", students_df.columns)
print("Marks DataFrame Columns:", marks_df.columns)

students_df.columns = students_df.columns.str.strip()
marks_df.columns = marks_df.columns.str.strip()
combined_df = pd.merge(students_df, marks_df, on='student_id', how='outer')

print("\nB: Print the maximum and minimum marks Gender wise")
max_min_marks_gender = combined_df.groupby('Gender')['Mark'].agg(['max', 'min'])
print("Maximum and Minimum Marks Gender wise:")
print(max_min_marks_gender)

print("\nC: Print all the students IDs and their marks who have scored more than the average marks of the class")
average_marks = combined_df['Mark'].mean()
above_average_students = combined_df[combined_df['Mark'] > average_marks][['student_id', 'Mark']]
print("Students with marks above average:")
print(above_average_students)

print("\nD: Print the dataframe who are Males and are Employed")
males_employed = combined_df[(combined_df['Gender'] == 'Male') & (combined_df['Employed'] == 'yes')]
print("Males who are Employed:")
print(males_employed)

print("\n E: Create a new Column ‘IQ_level’ which will have 3 values (Intelligent, Mediocre, weak)")
def iq_level(score):
    if score > 80:
        return 'Intelligent'
    elif score >= 50:
        return 'Mediocre'
    else:
        return 'Weak'

combined_df['IQ_level'] = combined_df['Mark'].apply(iq_level)
print("\nCombined DataFrame with IQ_level:")
print(combined_df)

print("\n F: Count the number of males and females from each city")
gender_count_by_city = combined_df.groupby(['City', 'Gender']).size().unstack(fill_value=0)
print("Count of Males and Females from each city:")
print(gender_count_by_city)

print("\nG: Print the top 5 Male scorers")
top_5_male_scorers = combined_df[combined_df['Gender'] == 'Male'].nlargest(5, 'Mark')
print("Top 5 Male Scorers:")
print(top_5_male_scorers)

print("\nH: Replace Male with M and Female with F and export to CSV")
combined_df['Gender'] = combined_df['Gender'].replace({'Male': 'M', 'Female': 'F'})
combined_df.to_csv('D:/test.csv', index=False)
print("Combined DataFrame after replacing Gender values:")
print(combined_df)

print("\nI: Check if any student_ID is duplicated")
duplicates = combined_df[combined_df.duplicated(['student_id'], keep=False)]
print("Duplicated Student IDs:")
print(duplicates)

print("\nJ: Create a separate dataframe with all Integer/Float variables")
numeric_df = combined_df.select_dtypes(include=['int64', 'float64'])
print("Dataframe with Integer/Float variables:")
print(numeric_df)

print("\nK: Get those Student_IDs which are present in Students table but not in Marks table")
students_only = students_df[~students_df['student_id'].isin(marks_df['student_id'])]
print("Student IDs present in Students table but not in Marks table:")
print(students_only['student_id'])
```

```python
data = {
    'Alpha': [23, 212, None, 48],
    'Beta': [None, 92, 38, 42],
    'Gamma': [31, None, None, 74]
}
df = pd.DataFrame(data)
print("Sample Dataframe")
print(df)
df['Alpha'] = df['Alpha'].fillna(df['Alpha'].mean())
print("\nDataFrame after mean imputation for column Alpha:")
print(df)

df['Beta'] = df['Beta'].fillna(df['Beta'].median())
print("\nDataFrame after median imputation for column Beta:")
print(df)

df['Gamma'] = df['Gamma'].fillna(df['Gamma'].mode()[0])
print("\nDataFrame after mode imputation for column Gamma:")
print(df)

df['Alpha'] = df['Alpha'].ffill()  # Use ffill() instead of fillna(method='ffill')
print("\nDataFrame after forward fill for column Alpha:")
print(df)

df['Beta'] = df['Beta'].bfill()  # Use bfill() instead of fillna(method='bfill')
print("\nDataFrame after backward fill for column Beta:")
print(df)

df_dropped_rows = df.dropna(axis=0)
print("\nDataFrame after dropping rows with missing values:")
print(df_dropped_rows)

df_dropped_columns = df.dropna(axis=1)
print("\nDataFrame after dropping columns with missing values:")
print(df_dropped_columns)
```
