```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

#### India Food Analysis

```python
path = "C:/Data_Science/Data Visualisation/Indian_food.csv"
df = pd.read_csv(path)
```

a)	In the state column, replace the -1 with the mode of state column.

```python
mode_state = df['state'].mode()[0]
df['state'] = df['state'].replace('-1', mode_state)
```

b)	Whatever the state you got in previous question, fill the corresponding region in the region column.

```python
mode_region = df[df['state'] == mode_state]['region'].mode()[0]
```

```python
df['region'].replace('-1',mode_region)
```

c)	Replace the -1 with mean for cook time variable.

```python
mean_cook_time = round(df['cook_time'].mean())
df['cook_time']=df['cook_time'].replace(-1,mean_cook_time)
```

d)	Most preferred flavours by number of customers using pie chart.

```python
flavour_count = df['flavor_profile'].value_counts()
```

```python
color = ['#0000FF','#1E90FF','#6495ED','#87CEFA','#ADD8E6']
fig = px.pie(df,names='flavor_profile',color_discrete_sequence=color,title='Most preferred flavours distribution')
fig.update_layout(height=500,width=450,title_x=0.5)
fig.show()
```

e)	Most requested course by customers using Bar chart. Print the labels also on the top of each bar. Then sort the bar chart in descending order.

```python
course_counts = df['course'].value_counts().reset_index()
course_counts.columns = ['course', 'count']
# Create the bar chart
fig_bar = px.bar(course_counts.sort_values('count', ascending=False), x='course', y='count', color_discrete_sequence=px.colors.sequential.Viridis,title='Most Requested Course by Customers',text='count')
fig_bar.update_layout(height=400,width=500,title_x=0.5,xaxis_title='Course Name',yaxis_title = 'No. of Customers')
fig_bar.show()
```

f)	Comparison between Vegetarians and Non-Veg orders in each region. 

```python
diet_df=df.groupby(['region','Diet']).size()
```

```python
diet_df.columns=['region','diet','count']
```

```python
diet_counts = df.groupby(['region', 'Diet']).size().reset_index(name='count')

fig_bar = px.bar(diet_counts, x='region', y='count', color='Diet',
                 color_discrete_map={'Vegetarian': '#577f49', 'Non-Vegetarian': '#FF6347'},
                 title='Comparison of diet in each region', text='count')

fig_bar.update_layout(height=500, width=600, title_x=0.5, 
                      xaxis_title='Region')
fig_bar.show()
```

g)	Create the box plot for Order cooking time to see the distribution. Also check if there are any outliers in this column.

```python
fig = px.box(df, y='cook_time', title='Order Cooking Time Distribution')
fig.update_layout(height=500, width=500, title_x=0.5,yaxis_title='Cooking Time')
fig.show()
```

### Billionaires Stats

```python
path = "C:/Data_Science/Biilionaires Stats.csv"
billion_df = pd.read_csv(path,encoding='latin-1')
```

a)	Create a dataframe which will have the billionaires whose age is greater than or equal to 30. 
Then create a scatter plot to show the relationship between age and net worth.

```python
age_greater_30 = billion_df[billion_df['age']>=30]
```

```python
fig_scatter = px.scatter(age_greater_30,x="age",y="finalWorth",color="finalWorth",height=400,width=700,title='relationship between age and net worth')
fig_scatter.show()
```

b)	Create a histogram to see the distribution of ages.

```python
fig = px.histogram(billion_df, x='age', nbins=10, title='Distribution of Ages', labels={'age':'Age'},text_auto=True)
fig.update_layout(title_x=0.5)
fig.show()
```

c)	Create a heatmap and see the correlation between all the numeric variables. Print the correlation value also in the heatmap.
d)	Save the figure in your pwd generated through previous question.

```python
numeric_columns = billion_df.select_dtypes(include='number')

corr_matrix = numeric_columns.corr()

fig = px.imshow(corr_matrix, text_auto=True, color_continuous_scale='Viridis', title='Correlation Heatmap')
fig.update_layout(
    font=dict(
        family="Arial",
        size=12
    ),
    width=600,
    height=500,
    title_x=0.5
)
fig.show()

#for saving image
#pip install kaleido 
fig.write_image('correlation.png')
```
