```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
c
path = "/Users/satishmac/Downloads/indian_food.xlsx"
df = pd.read_excel(path)

# a) Replace -1 in the state column with the mode of the state column
mode_state = df['state'].mode()[0]
df['state'] = df['state'].replace('-1', mode_state)

# b) Fill the corresponding region in the region column for the mode state
mode_region = df[df['state'] == mode_state]['region'].mode()[0]
df['region'] = df['region'].replace('-1', mode_region)

# c) Replace -1 with the mean for the cook_time variable
mean_cook_time = round(df['cook_time'].mean())
df['cook_time'] = df['cook_time'].replace(-1, mean_cook_time)

# Display the cleaned dataset
print(df.head())
```

```python
#d) Most Preferred Flavours (Pie Chart)
# Count the occurrences of each flavour
flavour_count = df['flavor_profile'].value_counts()

# Create a pie chart using Plotly
fig = px.pie(flavour_count, names=flavour_count.index, values=flavour_count.values,
             title='Most Preferred Flavours Distribution')
fig.update_layout(height=500, width=500, title_x=0.5)
fig.show()
```

```python
#e) Most Requested Course (Bar Chart)
# Count the occurrences of each course
course_counts = df['course'].value_counts().reset_index()
course_counts.columns = ['course', 'count']

# Create a bar chart using Plotly
fig = px.bar(course_counts.sort_values('count', ascending=False), 
             x='course', y='count', text='count',
             title='Most Requested Course by Customers')
fig.update_layout(height=400, width=600, title_x=0.5,
                  xaxis_title='Course Name', yaxis_title='Number of Customers')
fig.show()
```

```python
#f) Comparison Between Vegetarians and Non-Veg Orders (Stacked Bar Chart)
# Group by region and diet, then count the occurrences
diet_counts = df.groupby(['region', 'Diet']).size().reset_index(name='count')

# Create a stacked bar chart using Plotly
fig = px.bar(diet_counts, x='region', y='count', color='Diet',
             color_discrete_map={'Vegetarian': '#577f49', 'Non-Vegetarian': '#FF6347'},
             title='Comparison of Diet in Each Region', text='count', barmode='stack')
fig.update_layout(height=500, width=700, title_x=0.5,
                  xaxis_title='Region', yaxis_title='Number of Orders')
fig.show()
```

```python
#g) Order Cooking Time Distribution (Box Plot)
# Create a box plot for cook_time using Plotly
fig = px.box(df, y='cook_time', title='Order Cooking Time Distribution')
fig.update_layout(height=500, width=500, title_x=0.5, yaxis_title='Cooking Time (minutes)')
fig.show()
```

```python
# Load the Billionaires Dataset dataset
path = "/Users/satishmac/Downloads/Billionaires_Stats.xlsx"
billion_df = pd.read_excel(path)

# Display the first few rows
print(billion_df.head())
```

```python
#a) Scatter Plot: Age vs. Net Worth
# Filter billionaires with age >= 30
age_greater_30 = billion_df[billion_df['age'] >= 30]

# Create a scatter plot using Plotly
fig = px.scatter(age_greater_30, x='age', y='finalWorth', color='finalWorth',
                 title='Relationship Between Age and Net Worth')
fig.update_layout(height=400, width=700, title_x=0.5,
                  xaxis_title='Age', yaxis_title='Net Worth')
fig.show()
```

```python
#b) Histogram: Distribution of Ages
# Create a histogram using Plotly
fig = px.histogram(billion_df, x='age', nbins=20, 
                   title='Distribution of Ages', labels={'age': 'Age'})
fig.update_layout(height=400, width=600, title_x=0.5,
                  xaxis_title='Age', yaxis_title='Count')
fig.show()
```

```python
#c) Heatmap: Correlation Between Numeric Variables
# Select numeric columns
numeric_columns = billion_df.select_dtypes(include='number')

# Compute the correlation matrix
corr_matrix = numeric_columns.corr()

# Create a heatmap using Plotly
fig = px.imshow(corr_matrix, text_auto=True, color_continuous_scale='Viridis',
                title='Correlation Heatmap')
fig.update_layout(height=500, width=600, title_x=0.5)
fig.show()

# Save the heatmap as an image
fig.write_image('correlation_heatmap.png')
```
