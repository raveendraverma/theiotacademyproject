```python
students = pd.read_excel('students.xlsm')
marks = pd.read_excel('marks.xlsm')
print(students)
print(marks)
```

## a)	Combine both the dataframes into single dataframe which will have all the records from both the tables.

```python
combined_data = pd.merge(students, marks, on='Student_id', how='outer')
print(combined_data)
```

## b)	Print the maximum and minimum marks Gender wise.

```python
gender_marks = combined_data.groupby('Gender')['Mark'].agg(['max', 'min'])
print(gender_marks)
```

## c)	Print all the students IDs and their marks who have scored more than the average marks of the class.

```python
average_marks = combined_data['Mark'].mean()
above_avg_students = combined_data[combined_data['Mark'] > average_marks][['Student_id', 'Mark']]
print(above_avg_students)
```

## d)	Print the dataframe who are Males and are Employed.

```python
males_employed = combined_data[(combined_data['Gender'] == 'Male') & (combined_data['Employed'] == True)]
print(males_employed)
```

## e)	Create a new Column ‘IQ_level’ which will have 3 values (Intelligent, Mediocre, weak). If student scored than 80 then Tag him as Intelligent, if student scored between 50-80, then Mediocre, else weak.

```python
def iq_level(score):
    if score > 80:
        return 'Intelligent'
    elif score >= 50:
        return 'Mediocre'
    else:
        return 'Weak'

combined_data['IQ_level'] = combined_data['Mark'].apply(iq_level)
print(combined_data[['Student_id', 'Mark', 'IQ_level']])
```

## f)	Count the number of males and females from each city.

```python
gender_city_count = combined_data.groupby(['City', 'Gender']).size().unstack(fill_value=0)
print(gender_city_count)
```

## g)	Print the top 5 Male scorers.

```python
top_5_male_scorers = combined_data[combined_data['Gender'] == 'Male'].nlargest(5, 'Mark')
print(top_5_male_scorers[['Student_id', 'Mark']])
```

# 3. Explain the concept of missing values. How can you identify the missing values in a Pandas DataFrame ?
# What are the different ways of treating/Imputing/Deleting the missing values.
# Explain with example.


```python
data = {'Name': ['Raghul', 'Rajesh', 'Ram', 'Ravi', 'Raja','Rakshan'],
        'Age': [25, 49, np.nan, np.nan, 37, 22],
        'Marks': [98, 78, 65, np.nan, 52, np.nan]}

df = pd.DataFrame(data)
print(df)
```

### We can identify missing values using the function df.isnull(). It will return boolean dataframe indicating the missing values

```python
print(df.isnull())
```

```python
print(df.isnull().sum())
```

### i. Deleting rows or columns with missing values

```python
df_dropped_rows = df.dropna()
print(df_dropped_rows)
```

```python
df_dropped_cols = df.dropna(axis=1)
print(df_dropped_cols)
```

### ii. We can imputate by filling real values.

```python
df_filled = df.fillna(20)  # Filling a fixed value of 20 instead of missing values
print(df_filled)
```
