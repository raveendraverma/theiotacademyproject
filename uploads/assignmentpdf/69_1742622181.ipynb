```python
'''Q3. Write a Python user defined function that takes a list of 
numbers from user and returns a dictionary with the number as the 
key and its square as the value.'''

def squared(l): #l is input list
    squared_dict={}
    for i in l:
        squared_dict[i]=i*i
    return squared_dict

squared([1,2,3,4,5,6])
        
```

```python
'''Q8 . Write a Function to take a list from user and return 2 lists:
-	1st list should contain all even indexed numbers.
-	2nd list should contain all odd indexed numbers'''

def sorted(l):
    even_list=[]
    odd_list=[]
    for i in range(len(l)):
            if i%2==0:
                even_list.append(i)
            else:
                odd_list.append(i)
    print(even_list)
    print(odd_list)

sorted([1,2,3,4,8,6,5])
```

```python
'''Q9. Create a function longest_word(sentence) that takes a sentence as input and returns  the  longest word in  the  sentence. If two or  more words have the  same length, return the first one.

Eg: longest_word("Python programming is fun")   # Output: 'programming'''

def longest_word(l):
    l=list(l.split(" "))
    longest=""
    for i in l:
        if len(i)>len(longest):
            longest=i
    
    return longest

longest_word("Python programming is fun")

    

```

```python
'''Q10. Write a Python  program  that takes a list and counts  the frequency of each  element using a dictionary. The keys of the dictionary should be the list elements, and the values should be the counts'''
def count_instance(l):
    d={}
    for i in l:
        d[i]=l.count(i)
    return d

count_instance([1,2,2,3,4,5])
```
